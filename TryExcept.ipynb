# Odchytávání chyb v Pythonu (`try` / `except`)

Když programujete jednoduché úkoly, kód běží v ideálním prostředí. Vy víte, co do programu zadáváte, a víte, kde máte uložené soubory. Jakmile ale program nebo hru pustí do ruky **skutečný uživatel (hráč)**, začne chaos.

Hráči dělají chyby:

* Zadávají text tam, kde má být číslo.
* Mažou soubory se hrou, které mazat nemají.
* Odpojují internet uprostřed načítání.

Pokud na tyto situace kód nepřipravíte, Python při první chybě **vypíše ošklivou hlášku a celá hra okamžitě spadne na plochu.**

Abychom tomu zabránili, používáme bloky `try` (zkus) a `except` (pokud nastane chyba). Je to taková záchranná síť pro náš program.

---

## Jak vypadá pád programu (Bez záchranné sítě)

Představte si hru, kde dělíte nasbírané body nebo lektvary mezi členy týmu. Co když má ale hráč v týmu 0 lidí? V matematice je dělení nulou zakázané.

### Tento kód shodí program:


```python
body = 100
pocet_hracu = 0

# Python zde narazí na chybu ZeroDivisionError a okamžitě ukončí hru!
body_na_jednoho = body / pocet_hracu
print(f"Každý hráč dostane {body_na_jednoho} bodů.")
print("Tento řádek se kvůli pádu už nikdy nespustí...")
```


---

## Příklad 1: Ochrana před špatným vstupem (Zadávání čísel)

Nejčastější chyba: Chcete po hráči číslo, ale on na klávesnici napíše text. Funkce `int()` neumí převést slovo "deset" na číslo `10` a program kvůli tomu spadne (`ValueError`).

Využijeme `try` a `except`. Pythonu řekneme: *"Zkus udělat tohle, a pokud to vyvolá chybu, udělej náhradní plán."*

### Buňka s kódem:


```python
vstup_od_uzivatele = "padesat" # Předstíráme, že tohle hráč napsal

try:
    # 1. KROK: Zkus opatrně převést text na číslo
    vsazene_zlato = int(vstup_od_uzivatele)
    print(f"Úspěšně jsi vsadil {vsazene_zlato} zlaťáků.")

except ValueError:
    # 2. KROK: Pokud to nebylo číslo, program nespadne, ale skočí SEM
    print("Chyba: Nezadal jsi platné číslo! Systém automaticky sází 0 zlaťáků.")
    vsazene_zlato = 0

# Program žije dál
print(f"Pokračujeme ve hře. Vsadil jsi celkem: {vsazene_zlato}")
```

## Příklad 2: Chybějící nebo poškozený soubor s pozicí (`Save Game`)

Představte si, že vaše hra při startu hledá soubor `savegame.json`, aby načetla rozjetou hru. Jenže hráč hru hraje poprvé, takže žádný soubor ještě neexistuje. Bez `try/except` by nováček hru ani nespustil, protože by mu hned spadla s chybou `FileNotFoundError`.

### Buňka s kódem:



```python
import json

jmeno_souboru = "ulozeny_postup.json"

try:
    print("Načítám uloženou pozici ze souboru...")
    with open(jmeno_souboru, "r", encoding="utf-8") as soubor:
        data = json.load(soubor)
    print("Pozice úspěšně načtena. Vítej zpět!")

except FileNotFoundError:
    # Pokud soubor neexistuje, hra nespadne. Založíme nový profil!
    print("Soubor s pozicí nenačten (asi hrajete poprvé). Vytvářím nový profil na 1. úrovni.")
    data = {"jmeno_hrace": "Novacek", "level": 1, "zlato": 100}

# Hra bezpečně pokračuje s daty, která máme (buď načtená, nebo nová)
print(f"Hra spuštěna pro: {data['jmeno_hrace']} (Level: {data['level']})")
```

## Příklad 3: Univerzální štít pro nečekané situace

Někdy nevíte, jaká chyba se může stát (např. chyba grafické karty, výpadek sítě apod.). Pro tyto případy existuje univerzální zachytávač `Exception`. To je "matka všech chyb" v Pythonu. Zachytí úplně cokoli.

### Buňka s kódem:


```python
try:
    # Tady se může pokazit cokoliv divného
    # Simulujeme pokus o otevření prázdného listu
    inventar = []
    aktivni_zbran = inventar[5] # Hledáme šestou věc v prázdném batohu (IndexError)

except Exception as neznama_chyba:
    # Zachytí jakoukoliv chybu a uloží její text do proměnné 'neznama_chyba'
    print(f"🚨 Hra detekovala kritický problém: {neznama_chyba}")
    print("Aktivuji nouzový režim, abychom zachránili data před pádem.")
    aktivni_zbran = "Pěsti"

print(f"Hra pokračuje. Tvoje zbraň: {aktivni_zbran}")
```


---

## Shrnutí: Jak to funguje dohromady?

1. **`try:`** Sem dáváte kód, kterému stoprocentně nevěříte (vstupy uživatelů, práce se soubory).
2. **`except JmenoChyby:`** Tady řešíte konkrétní problém (např. špatné číslo, chybějící soubor).
3. **`except Exception as e:`** Tohle dáváte na konec jako poslední pojistku, která pochytá zbytek, aby program za žádnou cenu nespadl.
